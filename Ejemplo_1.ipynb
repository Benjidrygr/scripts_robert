```python
from ortools.sat.python import cp_model
```

```python
class ListarSoluciones(cp_model.CpSolverSolutionCallback):
  def __init__(self, variables): # <-- constructor
    super().__init__() # <-- llamar al constructor de cp_model.CpSolver...
    self.__variables = variables # atributo privado
    self.soluciones = [] # lista vacía

  def on_solution_callback(self): # sobreescritura del metodo
    solucion_actual = {} # diccionario
    for var in self.__variables: # para cada solución encontrada
      solucion_actual[var.Name()] = self.Value(var) # guardarlo

    self.soluciones.append(solucion_actual) # guardar la solución
    print(f'Solucion actual: {solucion_actual}')  # mostrar la solución
```

```python
model = cp_model.CpModel()
```

```python
# variables
x1 = model.NewIntVar(1, 2, 'x1')
# la varialbe x1 es de tipo entero y toma valores de 1 a 2
x2 = model.NewIntVar(1, 2, 'x2')
# la variable x2 es de tipo entero y toma valores de 1 a 2
x3 = model.NewIntVar(1, 3, 'x3')
# la variable x3 es de tipo entero y toma valores de 1 a 3
```

```python
# restricciones:
model.Add(3 * x1 + x2 + x3 == 10) # primera restriccion
# model.Add(x1 == x2) # son iguales (creará después una inconsistencia)
```

```python
# model.AddAllDifferent([x1, x2, x3]) # todos los valores deben ser distintos
```

```python
solver = cp_model.CpSolver()
solver.parameters.enumerate_all_solutions = True
callback = ListarSoluciones([x1, x2, x3])
status = solver.Solve(model, callback)
```
